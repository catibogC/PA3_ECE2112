```python
import pandas as pd
```

```python
cars = pd.read_csv('cars.csv')
```

## A. POSITIONAL AND LABEL BASED SLICING

```python
cars
```

```python
cars.shape
```

```python
cars_6_to_10 = cars.iloc[[0,5,6,7,8,9]]
cars_6_to_10
```

```python
column_selection = cars_6_to_10.loc[[0,5,6,7,8,9], ['Model', 'mpg', 'cyl', 'hp', 'gear']]
column_selection
```

## B. MODEL LOOKUP

```python
toyota_corolla = cars.loc[cars['Model']=='Toyota Corolla']
toyota_corolla
```

```python
pontiac_firebird = cars.loc[cars['Model']=='Pontiac Firebird', ['Model', 'mpg', 'hp', 'wt']]
pontiac_firebird
```

## C. MULTI-MODEL SUBSETTING

```python
models = ['Datsun 710', 'Lotus Europa', 'Ferrari Dino']
models
```

```python
columns = ['Model', 'mpg', 'cyl', 'hp', 'gear']
columns
```

```python
selected_cars = cars.loc[cars['Model'].isin(models), columns]
selected_cars
```
